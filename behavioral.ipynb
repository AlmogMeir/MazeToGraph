```python
def find_reward_sequences(reward_ports, start_timestamps, reward_1=1, reward_2=2, min_length=2):
    """
    Find sequences of consecutive trials that only have specified reward ports.
    Any other reward port (including None) breaks the sequence.
    
    Args:
        reward_ports: list of reward port numbers for each trial
        start_timestamps: array of trial start times
        reward_1: first allowed reward port number (default: 1)
        reward_2: second allowed reward port number (default: 2)
        min_length: minimum number of trials in a sequence (default: 2)
    
    Returns:
        sequences: list of dicts with 'start_trial', 'end_trial', 'start_time', 'duration', 'num_trials'
    """
    sequences = []
    current_sequence_start = None
    
    for i, port in enumerate(reward_ports):
        # Check if this trial is one of the allowed reward ports
        if port in [reward_1, reward_2]:
            # If we're not in a sequence, start one
            if current_sequence_start is None:
                current_sequence_start = i
        else:
            # If we encounter a different reward port, end the current sequence
            if current_sequence_start is not None:
                # Record the sequence
                start_trial = current_sequence_start
                end_trial = i - 1
                num_trials = end_trial - start_trial + 1
                
                # Only add sequence if it meets minimum length requirement
                if num_trials >= min_length:
                    start_time = start_timestamps[start_trial]
                    end_time = start_timestamps[end_trial]
                    duration = end_time - start_time
                    
                    # Count port changes within the sequence
                    port_changes = 0
                    for j in range(start_trial, end_trial):
                        if reward_ports[j] != reward_ports[j + 1]:
                            port_changes += 1
                    
                    sequences.append({
                        'start_trial': start_trial,
                        'end_trial': end_trial,
                        'start_time': start_time,
                        'duration': duration,
                        'num_trials': num_trials,
                        'port_changes': port_changes
                    })
                
                # Reset sequence start
                current_sequence_start = None
    
    # Handle the case where the last sequence extends to the end
    if current_sequence_start is not None:
        start_trial = current_sequence_start
        end_trial = len(reward_ports) - 1
        num_trials = end_trial - start_trial + 1
        
        # Only add sequence if it meets minimum length requirement
        if num_trials >= min_length:
            start_time = start_timestamps[start_trial]
            end_time = start_timestamps[end_trial]
            duration = end_time - start_time
            
            # Count port changes within the sequence
            port_changes = 0
            for j in range(start_trial, end_trial):
                if reward_ports[j] != reward_ports[j + 1]:
                    port_changes += 1
            
            sequences.append({
                'start_trial': start_trial,
                'end_trial': end_trial,
                'start_time': start_time,
                'duration': duration,
                'num_trials': num_trials,
                'port_changes': port_changes
            })
    
    return sequences

# Get start timestamps
start_timestamps = session_data.TrialStartTimestamp

# Find sequences (reward 1 and 2, minimum 2 trials)
sequences = find_reward_sequences(reward_ports, start_timestamps, reward_1=1, reward_2=2, min_length=2)

print(f"Found {len(sequences)} sequences of reward 1 and 2 trials")
print("\nSequence details:")
for i, seq in enumerate(sequences):
    print(f"\nSequence {i+1}:")
    print(f"  Trials: {seq['start_trial']} to {seq['end_trial']}")
    print(f"  Number of trials: {seq['num_trials']}")
    print(f"  Port changes: {seq['port_changes']}")
    print(f"  Start time: {seq['start_time']:.2f} s")
    print(f"  Duration: {seq['duration']:.2f} s")
    print(f"  Reward pattern: {reward_ports[seq['start_trial']:seq['end_trial']+1]}")
```
